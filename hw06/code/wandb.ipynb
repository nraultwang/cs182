```python
project = 'CS182 WANDB.AI Practice Notebook'
learning_rate = 0.01
epochs = 5
architecture ='CNN'
dataset = 'CIFAR-10'
batch_size = 256
# momentum = 0.9
log_freq = 2
print_freq = 50
cuda = torch.cuda.is_available()
device = torch.device("cuda" if cuda else "cpu")
device

hyperparameters = torch.linspace(1e-3, 1e-5, 5)
print(hyperparameters)
```

```python
def analyze_model_and_inputs(model):
    # print("Train data size: {}".format(len(data_train)))
    # print("Test data size: {}".format(len(data_test)))
    
    # Fetch an example image to get image size
    # image, label = data_train[0]
    # print("Image input size: {}".format(image.size()))

    # Get model parameter count
    print("Model parameters: {}".format(sum(p.numel() for p in model.parameters() if p.requires_grad)))

    # Get model size in MB
    print("Model size estimate (MB): {}".format(sum(p.numel() * p.element_size() for p in model.parameters()) / 1e6))

model = ResNet101()
model.to(device)  # Load the model into GPU memory
analyze_model_and_inputs(model)
```

```python
def train_with_lr(learning_rate, batch_size, model_arch):
    # create writer for TensorBoard
    run_name = f'{model_arch=}.{learning_rate=}.{batch_size=}'
    wandb.init(
        # set the wandb project where this run will be logged
        project=project,
        
        # track hyperparameters and run metadata
        config={
        "learning_rate": learning_rate,
        "architecture": model_arch,
        "dataset": dataset,
        "epochs": epochs,
        "batch_size": batch_size,
        # "momentum": momentum
        }
    )
    print(run_name)
    
    # Select model architecture
    if model_arch == 'mlp':
        model = MLP()
    elif model_arch == 'cnn':
        model = BasicConvNet()
    elif model_arch == 'resnet18':
        model = ResNet18()
    elif model_arch == 'resnet50-test4':
        model = ResNet50()
    elif model_arch == 'resnet101':
        model = ResNet101()
    model.to(device)

    # Initialize training infrastructure
    loss_fn = torch.nn.CrossEntropyLoss()
    optimizer = optim.AdamW(model.parameters(), lr=learning_rate)

    # num_data_workers = os.cpu_count() // 2 if os.cpu_count() else 4
    num_data_workers = 4
    print(f"Using {num_data_workers} data workers.")
    
    # Initialize trainloader
    trainloader = torch.utils.data.DataLoader(
        dataset=trainset,
        shuffle=True,
        batch_size=batch_size,
        num_workers=num_data_workers,
        pin_memory=True
    )

    # writer = SummaryWriter(comment=run_name)
    for epoch in range(epochs):  # loop over the dataset multiple times
        running_loss = 0.0
        running_acc = 0.0
        for i, data in tqdm(enumerate(trainloader), unit=' batch'):
            # extract input + label tensors for this batch
            inputs = data[0].to(device, non_blocking=True)
            labels = data[1].to(device, non_blocking=True)
    
            # reset gradients for this batch
            optimizer.zero_grad()
    
            # make predictions for this batch
            outputs = model(inputs)
    
            # compute loss between ground truth and predicted labels
            loss = loss_fn(outputs, labels)
    
            # do backprop
            loss.backward()
    
            # adjust learning weights
            optimizer.step()
        
            # When you want to log a value, use the writer. When adding a scalar, the format is as follows:
            # add_scalar(tag, scalar_value, global_step=None, walltime=None, new_style=False, double_precision=False)
            # if batch_idx > 20:
                # writer.add_scalar('Training Loss', loss.item(), global_step=batch_idx)
            accuracy = torch.mean((torch.argmax(outputs, dim=1) == labels).float()).item() * 100
    
            # print statistics
            running_acc += accuracy
            running_loss += loss.item()
            if i % log_freq == log_freq - 1:
                wandb.log({'accuracy': accuracy, 'loss': loss.item()})
                
            # if i % print_freq == print_freq - 1:    # print every 2000 mini-batches
        print(f'[{epoch + 1}, {i + 1:5d}] loss: {running_loss / i:.5f} accuracy: {running_acc / i:.5f}')
        running_loss = 0.0
        running_acc = 0.0
    wandb.finish()
    
    # Finally, when you are done logging values, close the writer
    # writer.close()
    # raise NotImplementedError

def run():
    for lr in hyperparameters:
        train_with_lr(lr, batch_size=2048, model_arch='resnet50-test4')
    
# train_with_lr(hyperparameters[0], batch_size=128)
run()
```
